### Don't Forget the Metadata

Downloading data isn't our ultimate goal. We want to generate, for different criteria, lists of interesting exoplanets at which we can aim our expensive telescope. And since the exoplanet data is continually updated, we should capture the provenance of the data.

Here's a prompt that could help:

> **PROMPT:** When run, the program should also create metadata describing its work. This metadata should include: the date and time of retrieval, the source table, the exact query, the selected fields, a hash of the response, and a listing of any transformations performed on the retrieved data. Tell me how you stored this metadata.

Again, the prompt lets the AI decide how to store the **metadata (i.e., data about the data)**. It hopefully will choose JSON, and if not, we'll fix it.

The prompt also assumes a few things:

-   AI will know what program we want edited. If you stay in the same chat session, it should.
-   AI will know what we mean by "hash of the response." This is us thinking ahead. When we grab new data from the NASA Exoplanet Archive, we'll want to check if it is different from the data we've already stored. If it isn't, there's no use in saving this new download in our database. If you don't have any idea of what it means to take a hash of some data and why it's helpful, ask AI or come talk to a member of the teaching staff.
-   AI may transform the data when it stores it. It may not, but we don't want it to be silent about such transformations. This is something that you learn from experience.

Prompt the AI and check the changes it proposes to make to the program. When you like the changes, run the program and see how it works. Fix things as necessary.

### ACTIVITY #2: Pick a Database and Schema

Most of us (me included) are not database experts. There are lots of choices, and writing a sensible schema for our database is a task that sounds important and detailed. Unlike most of us, AI was trained on lots of information about database choices and has seen lots of example schemas. Let's ask it for advice.

Here's a prompt that might work:

> **PROMPT:** I would now like to write a Python program that stores the data from the generated JSON files in a relational database. Can you suggest a specific database to use and a schema for the stored data?

If the AI generates code along with its answer to our question, this is **a place where we probably don't want to look too deeply at that code**. Why? The code will be well beyond our understanding. Our sanity check would be better done at the level of the explanation that we asked the AI to provide. If the AI doesn't simultaneously produce a program, you should ask it to do so in a subsequent prompt. 

**The answer we want:**

-   The AI should recommend the **SQLite** database, which we can easily use via Python's built-in `sqlite3` library.
    - SQLite is a simple solution requiring just one file (e.g., `exoplanets.db`). It will be a binary file that you cannot open and read (as you can with a text, CSV, or JSON file).
    - SQLite doesn't require a separate database server. Our Python program can easily read and update this database file.
-   The AI-generated schema should define several tables and some indices.
    - The indices are a performance enhancement, which we'll ignore.
    - The AI should create a planets table and a separate host stars table. In this way, it can group the properties shared by several planets into the stars table.
    - The AI may create several tables for the metadata associated with the retrievals. When we built our solution, AI created separate tables for the fields and transformations. These data may be identical across several retrievals, and since (for example) the list of retrieved fields is quite large, the database shouldn't repeat this list in each retrieval row.

**Bottom line:** We want everyone in the same design space. This isn't a class on databases, and so you don't need to be too critical of the AI-generated schema. Just make sure the AI picks SQLite (and force it to if necessary) and a schema not too far from what's described above.

Prompt the AI and fix things as necessary. Make sure that this load-to-SQLite program runs on the JSON files produced by the program from Activity #1.
