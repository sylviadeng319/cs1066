You should know what the `cd` command does by now. About the `rm` command:

-   it removes the files listed
-   `attempt*` is a pattern that is matched against the files in the working directory (which you set with `cd`) that start with the word "attempt"
-   `-rf` are options on the `rm` command
    - `r` says that this command should work _recursively_, meaning that if the pattern matches a directory, then the command should re-run in that directory (normally, `rm` doesn't delete directories)
    - `f` says that this command should _forcibly remove_ any matches, which overrides any prompting you might have enabled (see the `-i` option on `rm`)
