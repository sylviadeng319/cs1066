### Migrate to a New Database and Schema

Your database as currently organized doesn't store the candidate exoplanet lists. A requirement of our problem-to-be-solved was that we "preserve each list with the corresponding human-generated notes, decisions, and follow-up status (i.e., its provenance)."

Your task:

-   Change your program that makes ranked view lists so that it also captures notes about the list (in addition to the provenance of the list).
-   Have AI help you to add a table with rows for each candidate list to your database. This will require a database migration:
    1. Create a new database with a schema that combines the old database's schema with this new table.
    2. Migrate the data in the old database into the new one.
    3. Add your current list to the new database.

### Commit Your Notes and Edits

Don't forget to review, commit, and push your edits so that you've updated your GitHub repo with today's work.

*   `git status` to review what's new and what's been edited.
*   `git add <list_of_files>` to indicate what you want committed.
*   `git commit -m "<your_commit_message>"` to commit the changes to the codespace's filesystem.
*   `git push` to push those committed changes to your GitHub repo.
*   `git status` to verify that you haven't forgotten anything.
