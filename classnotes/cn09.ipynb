### Refinement Strategies

As you iterate on the app and add filters on the data, practice and add to this list of strategies that'll help you to get better results with AI:

1.  **Add constraints.** Your first prompt should be purposefully vague. Above, I said that the AI was to "prioritize people by compensation." The earlier AI interpreted this as "lowest first" and the later as "highest first." If a constraint matters, be specific:

    - Don’t: "prioritize people by compensation"
    - Do: "prioritize people by highest compensation first -- you want to retain the most valuable talent"

    Constraints can include:

    - Direction: "highest first" vs. "lowest first"
    - Defaults: "default to X, but allow changing to Y"
    - Exclusions: "don't include a color picker" or "remove the bottom graph"
    - Boundaries: "the slider should range from 1 to the total number of employees"


2.  **Provide examples or references.** When asking for design changes, abstract descriptions like "modern" or "professional" are interpreted differently by everyone---including an AI.

    - Don’t: "make it look professional"
    - Do: "make it look like a Harvard website"

    Best practice: Attach a screenshot of what you want. AI can extract the styles.

3.  **Describe the problem, not just the symptom.** When an aspect of the design falls short, explain _why_ it isn't sufficient.

    - Don’t: "fix the sidebar"
    - Do: "The left side is completely unreadable because the text and background are both white"

    Explaining the root cause helps the AI make the right fix instead of a random one.
