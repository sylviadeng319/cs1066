### The Other Type of Loop

Open your Python program from the last activity in the IDE's editor window. It should contain a number of loops. See if you can find one. Ask if you're confused.

What you find is likely a for-loop, i.e., it includes the keyword `for`. As we discussed previously (and don't worry if you have forgotten the details), the structure of the for-loop is:

```python
for iteration_variable in iterable:
    loop_body
```

`for` and `in` are keywords; they must appear as written. `iterable` is a collection or sequence. It contains a fixed number of items. It is over these items that the loop iterates. The `iteration_variable` names each of these items, one at a time, on each iteration of the loop.

Examples:

-   Your program uses a for-loop because there is a fixed and known number of rows in the input CSV table. Each row must be processed. When every row is processed, you're done. This is a work list. The work list is constructed and then each item in the work list is completed.
-   When we dealt with large-but-straightforward problems at the start of this class, we decomposed problems into tasks and then completed each task. This is a kind of work list, and the process we described was for-loop-like.

More recently, we've done a different kind of iteration. For example, we kept iterating with the AI until the program it produced did what we wanted without any significant failings. **We stopped when the AI-generated program was good enough.**

In this work, we're still looping (iterating), but we don't have a fixed task list. We don't know how many times we'll iterate. We just know when we can consider our work done. For each loop iteration in this new kind of loop, we're checking a specific stopping condition (e.g., is the AI-generated program good enough?).

In Python (and many other programming languages), this is called a while-loop.

----

### The Syntax of a While-loop

```python
while continuation_condition:
    loop_body
```

`while` is a keyword. The `continuation_condition` has the opposite sense of the stopping condition mentioned above. It says that the loop should iterate again until this condition evaluates to `False`. The loop body had better do something to eventually change this condition to `False`, or you'll get an infinite loop (i.e., one that never stops iterating).

----

### While-loop Examples

What do you think this code does? Guess and then run it.

````python
i = 1

while i < 4:
    print(i)
    i = i + 1
````

### Really There's Only One Loop

A for-loop is just a special kind of while-loop. What does the following print?

````python
for i in range(1, 4):
    print(i)
````

That's right. Every for-loop can be written as an equivalent while-loop. Technically:

-   You pull out the work list's first element and assign it to the iteration variable.
-   The continuation condition checks to see whether the iteration variable has gone off the end of the work list.
-   There's some statement inside the loop body that moves to the next item on the work list before the end of the loop body.

Don't worry if you didn't follow those technical details that show the equivalence of for- and while-loops. Just remember:

-   Loops are important in programs. That's where a lot of work gets done.
-   There are two kinds of loops.
    - One (the for-loop) is for work-list processing. You have a number of tasks to do and you need to complete them one at a time.
    - The other (the while-loop) is more general. You want to keep doing the loop's work until some stopping condition is satisfied, which in a for-loop is that there aren't any more tasks in the work list!
-   A while loop is necessary when you do not know in advance how many times you'll need to iterate. If you see a while-loop in AI-generated code, you can bet that it requires a stopping condition that _needs to be checked at run time_.

Look for this difference the next time you're reading AI-generated code for gist.
