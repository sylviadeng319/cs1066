### Data Provenance

A working database is constantly ingesting, updating, deleting, and combining data. People who use databases typically aren't just interested in the current state of the data, but also its **provenance**.

Data provenance involves maintaining information on:

-   Origin: Who created the data and how did it come into being?
-   History: What modifications were made to the data, and when?
-   Ownership: Who made those modifications?

With data provenance, you can **prove authenticity** and **generate trust**. It is often used in compliance situations, and it is gaining importance in the training of AI models.

### A Simple Query

To retrieve data from a database, you write a **query**.

Let's say you'd like to see the names and orbital periods of some planets in the Catalog's Planetary Systems (`ps`) table. We can run this query manually with the URL:

https://exoplanetarchive.ipac.caltech.edu/TAP/sync?query=select+pl_name,pl_orbper+from+ps&format=json

Breaking down the URL:

-   You understand `https://` and `exoplanetarchive.ipac.caltech.edu`.
-   `/TAP/sync` asks the server to run the TAP query synchronously (i.e., immediately).
-   `?query=` means that the following is an ADQL query, which is an astronomy-oriented dialect of SQL.

> Terminology
>
> -  **SQL** stands for Structured Query Language. It is a programming language used with relational databases. It is a standard.
> -  A **relational database** organizes data into tables (like CSV files).
> -  **Queries** are turned into SQL, which the database understands.
> -  Knowing SQL enables you to work with many **Database Management Systems (DBMS)** such as MySQL, PostgreSQL, Microsoft SQL Server, and Oracle. Each vendor adds its own extensions to the language.

-   `select+pl_name,pl_orbper+from+ps` is the ADQL query encoded for placement in a URL. It basically says to select the two named columns from the `ps` table.

```python
SELECT pl_name, pl_orbper FROM ps
```

-   `&format=json` requests the output in JSON. The service supports other choices.

### ORM Query (EXTRA)

If you talk to a software engineer, they might mention **Object-Relational Mapping (ORM)**, which is a software technique that:

-   lets you write a database query in an object-oriented programming language like Python
-   translates it into standard SQL
-   sends the request to the specified database, and
-   transforms the returned database rows into (Python) objects.

Obvious and Non-obvious Benefits:

-   **Easier programming.** Write only Python or JavaScript rather than a mix of that language and SQL.
-   **Better security.** Translation into SQL automatically sanitizes inputs.
-   **Hides vendor-specific extensions.** Automatically handles any syntax variations.

Tradeoff: An ORM can generate inefficient and slow SQL for some highly complex tasks.

ORM libraries for Python:
-   [Django ORM](https://www.djangoproject.com/)
-   [SQLAlchemy](https://www.sqlalchemy.org/)

### Database Schema

A schema is a blueprint. It defines how data is organized in the relational database. It includes the:

-   table names
-   column/field names
-   datatypes for the fields
-   rules that govern the data
-   relationships that exist in the database

We expect:

-   The data in the database to change **frequently**.
-   The database schema to change **infrequently**.

Here's a description of these two tables:

-   `ps` (Planetary Systems): Contains one row per planet per reference/publication. It summarizes the values for the planet as given by that specific reference/publication. This also means that the data in any single row is **self-consistent**.
-   `pscomppars` (Planetary Systems COMPosite PARameterS): Contains one row per planet. It is useful for work where you're interested in the overall exoplanet population. The row of values for any particular planet will typically have fewer missing values than any row for that planet in `ps` because the `pscomppars` row pulls from multiple references. As such, the data in any row is not necessarily self-consistent.
