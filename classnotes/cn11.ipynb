### Today's Philosophical Musing

We've started to focus on assumptions, but not all assumptions are the same. The following is a 3-level categorization that you might keep in mind while working with AI.

1.  **A reasonable working assumption**. A low-stakes belief that helps you move forward. It was probably verified at some earlier point.

2.  **An unexamined assumption**. A belief that you treat as a fact, but you never verify as true.

3.  **A false certainty.** A belief that has grown in scope as early confirmations have led you to stop checking.

Adapted from [this article](https://pmctraining.com/articles-and-resources/assumptions-in-decision-making/).

### Is My Data Clean?

Given a data room, we used AI to get to a pleasing, modern-looking interface and help us wade through the myriad of decisions involved in a complex evaluation problem. In this work, we assumed that we had _clean_ data. 

What is clean data?

1.  **No missing values.** Every field contains a value.
2.  **No invalid values.** Every field adheres to a column's constraints. E.g., there's not a string where you expect an integer, or a value out of range.
3.  **No input errors.** E.g., there are no spelling mistakes in string values or mis-entered numbers in numeric fields.
4.  **No duplicated observations.** An often-discovered type of input error is a duplicated observation.
5.  **No inconsistent formatting.** Every value in a column is coded in the same format, e.g., every date is written as MM/DD/YYYY.
6.  **No data inconsistencies.** Data values in two observations don't contradict. E.g., two different temperature measurements supposedly taken at the same location at the same time.

In general, clean data exhibits these criteria: validity, accuracy, completeness, consistency, and uniformity.

----

### What is Data Cleaning?

Previously, we assumed and acted as if we had clean data, and we did work with largely clean data sets. But real-world data sets often aren't. They're messy (not a technical term).

> Before using any data set, you must _inspect_ and _clean_ it.

Wikipedia defines [data cleaning](https://en.wikipedia.org/wiki/Data_cleansing) as "_detecting_ incomplete, incorrect, or inaccurate parts of the data and then _replacing, modifying, or deleting_ the affected data." \[emphasis added]

AI can help with this (perhaps difficult and unexciting) task.

### What is an Outlier?

Most companies have levels in a category like "Computer and Information System Managers". We looked at chief and director, but this includes designations like junior, mid, senior, and staff. The BLS data often doesn't distinguish between these different levels.

This means that you get false "outliers" if you compare the salary of a senior engineer to the median salary of the broad bucket of "Software Developer".

### Understanding an LLM's Context Window

It's time to get technical about LLMs again. But don't worry, we'll keep our discussion at a high level, focusing on what you need to know to be a better tool builder and problem solver.

By now, you've seen that **iteration is a powerful technique** for moving from an initial idea to a truly effective and useful program. As the tools that we built have become more complex, your iteration cycles with AI have gotten longer. The introduction of messy data added even more.

At the start of a project, you might say that the AI:

-   feels sharp
-   remembers to cover everything in your prompt
-   makes crisp, clean code changes.

However, toward the end of a particularly long iteration cycle, the AI begins to:

-   misunderstand what you mean
-   repeat old bugs
-   forget constraints you set earlier
-   edit the wrong file
-   miss your manual edits.

AI companies are aware of this and are introducing techniques to handle this phenomenon. But you can avoid this degradation by **regularly starting a new chat**.

-   Think about it like paragraphs and sections in a paper. Each prompt starts a paragraph and the AI finishes it. You provide the topic sentence and supporting details. AI gets you to the concluding sentence.
-   You and AI string paragraphs together, and hopefully, this gets you someplace interesting.
-   But eventually, it's time for some structure, to start a new section. Often, the next section is related to the previous, and as you start that next section, you explain to the reader what's important to take from that previous section.

----

### The Why Behind This Behavior

Earlier, we talked briefly about how LLMs take a prompt and generate a coherent continuation of that text, one word (token) at a time. The model's training set was the basis for its knowledge (it produced the learned probabilities represented by billions of parameters in these models).

What makes these models work well is an _attention mechanism_. It focuses the generation mechanism on the important context in the prompt.

> For more technical details, if you're interested, see ["Deep dive into Text Generation Inference with LLMs" on huggingface.co](https://huggingface.co/docs/course/chapter1/8). I'll just use their simple example.

What is important in the following sentence? What's crucial to get to the next word, "Paris"?

> The capital of France is

As the Hugging Face tutorial says, "there have been significant advancements in scaling neural networks and making the attention mechanism work for longer and longer sequences, at lower and lower costs."

The **context window** is a model's **attention span**.

> If you overfill the window, the model's attention wavers, and like us, it starts making mistakes.

----

### What Is Considered Context?

In a simple chatbot, the context window starts with the prompt. But in the IDE's AI, it also includes the files in your project and the terminal's history. 

When you type a second prompt, say to fix a problem with the AI-generated code produced by the first prompt, this new prompt adds to the context, which you likely think of as a conversation—or chat history.

In a long conversation, this conversation will eventually contain:

-   old requirements that are no longer true
-   outdated assumptions
-   previous code versions referenced in messages
-   contradictions ("use pandas" / "remove pandas").

This dreck makes the context messy. The attention mechanism is overwhelmed.

----

### Starting Anew

Starting a new chat resets the context window. The AI no longer has any of the context from the previous conversation, except what appears in the project's current code.

So, starting a new chat means that you should carefully consider what context the AI needs from the terminated conversation. This includes: 

-   Re-stating your goal.
-   Listing what the AI shouldn't change (e.g., "keep existing UI behavior unchanged").
