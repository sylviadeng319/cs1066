### Today's Philosophical Musing

Last time we worked with our AI assistant to get to a pleasing, modern-looking interface. This time we tackle complexity and the assumptions that creep into your tool's code because AI has to convert your requests into concrete numbers and expressions.

Recall that we were given a data room full of information. We started in the "people" folder, and even there, we had an employee roster with extensive data on each individual. But all our first apps did was basically sort the employees on a single column (compensation), and it is well known that compensation alone isn't a measure of an individual's importance. In an acquisition, you decide whom to keep based on lots of factors: tenure, level, technical expertise, managerial acumen, and so forth.

We could figure out exactly how an app should sort the employees based on some complex computation, but there's almost certainly no single computation that is "right" for every startup acquisition. That's why we wanted a tool that would let us explore the space of possibilities with this data room.

To get to this flexible tool without following a Waterfall approach, we will need to allow the AI to make some _consequential design decisions_ for us.

To this point, we've used AI to generate code (of course), but also to make some fairly minor design decisions:

-   the look of the tool's interface
-   the libraries used behind the scenes

We sometimes changed these decisions when we wanted a different look or thought another library would be more effective or efficient. But importantly, the tool was working before the change and continued to produce the same basic answer after the change.

From a process point of view:

1.  We wrote a prompt specifying what we want in a tool.
2.  We checked the AI-generated code and iterated until the tool worked as desired.
3.  We then iterated some more to change the look of the tool.
4.  Finally, we used the tool to get our work done.

Today, we're going to stick an iteration loop into step 1, since we'll start prompting (asking AI to generate code) before we have a complete specification. In effect, we're working with the AI to search for the tool we want and permitting it to **fill in missing parts of the specification**, which are consequential design decisions!

In order for this work to go smoothly, it will be up to us to determine **whether the AI-generated design decisions make sense for what we need done**. And when they do, we can continue on with steps 2–4 above.

This is still checking and testing, but of a different kind.

### Really Assessing Whom To Keep

With our current problem-to-be-solved, our goal is to generate an _impact score_ for each of the startup's employees, and it is this score that will dictate the ranked list of individuals.

The impact score is a computation composed of multiple factors, each with a weighting. Here's a list of factors we can pull from `employee_roster.csv`.

-   **Tenure**: This specifies how long an employee has been at the company. Longer tenures can indicate that this person has deep knowledge of the market or the technology.
-   **Level**: This is tied to a person's seniority. It indicates their level of experience beyond their time at the startup.
-   **Department** and **Role**: These describe the individual's work. Successful companies have strong contributors in multiple areas.
-   **Number of Reports**: This number can correlate with an employee's managerial skill and organizational knowledge.

We'll combine these 4 factors with compensation to create an impact score, and to give you maximal flexibility, the AI-generated tool should let you adjust the weights of these factors.

### What is a dataframe?

Behind the scenes, in every tool we've had it build lately, the AI has looked at the CSV file and decided that it should use a data structure called a _dataframe_.

A dataframe is a two-dimensional, table-like object that organizes data into rows and columns, like a spreadsheet in Microsoft Excel or Google Sheets. It is the foundational tool for data manipulation and analysis in Python. Most programmers manipulate dataframes using tools in the Pandas library, which `people1.py` imports as `pd`.

### From people2 to people3

We're **debugging** with AI. In writing prompts while debugging, **be explicit with context and error information**.

> **### BEST PRACTICE IN DEBUGGING ###**
> 
> -   Copy the Python error message and include it in your prompt.
> -   Provide an explanation of what you did that led to the error. Be as complete as possible. Imagine that you were helping the AI to reproduce the error.
> -   Ask the AI to explain what went wrong so that you can compare your understanding of the problem with the AI's.
>     - If your diagnosis and the AI's agree, you're both probably on the right track.
>     - If the diagnoses disagree, one of you is wrong, and it's important to figure out which.
> -   Request a fix.

### Regression Testing

We've focused recently on looks and design. We've made progress toward a useful tool, and we've remembered to checkpoint our work. **Checkpoints are also a great place to create test suites.**

With every new feature you add, you risk silently breaking the tool's existing functionality. While you can extract yourself from unproductive or undesirable changes with version control, it’s not always obvious that a subtle mistake with an _existing_ feature has crept into your tool as you introduce _new_ features.

**Regression tests help you to quickly and automatically identify such problems.**

In this example, the error was visibly displayed. But you can't rely on catching all errors by eye, especially since AI outputs often look reasonable at a glance:

-   Numbers are formatted correctly.
-   The UI works.
-   The app doesn't crash.

As the tools you build become more complicated, it’s also unreasonable to expect that you’ll check every calculation the program does and every number it prints. This type of manual checking is mind-numbing. 

Furthermore, when you add a new feature, you'll naturally focus your checks and tests on that new feature, not the existing features.

In short, manual checking:

-   doesn't scale as your tool takes on more work
-   goes against human nature as you iteratively add features

Regression tests automate away the need to manually check whether the new version unexpectedly breaks any of the existing features. They are a staple of the world of software engineering, and they have become even more important with AI-generated code.

We've already talked about building tests with `pytest`. Regression testing involves:

-   Running these tests on each new version before you commit it.
-   Adding to these tests as you implement new features and capabilities into your tool.
