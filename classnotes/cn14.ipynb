### Improvements Often Found in RAG Systems

**Filtering the Final List**

The retriever might filter the produced lists to remove documents not appropriate for user.

-   E.g., when a corporate user in one department shouldn't see the documents in another department.

**Chunking Choices**

Simplest approach: chunk a document by a fixed number of characters.

Tension:

-   Too large a chunk (e.g., whole file) provides full context but hard to identify where the match occurs.
-   Too small a chunk (e.g., 50 characters) directly identifes reference but without much context.

Improvements:

-   **Chunking on particular characters** (e.g., newlines in text documents or function boundaries in code). Uses natural boundaries in the content for chunk boundaries.
-   **Overlapping chunks**, typically expressed as an overlap percentage (e.g., chunk every 500 characters but overlap the first and last 50 characters). Diminishes the importance of choosing a good chunk boundary.
-   **Semantic chunks**, in which you preprocess the document to identify sensible chunks.
-   And others.

**Query Parsing**

Challenge: Unsurprisingly, most human-generated prompts make for bad search queries.

Solution: Retriever parses prompt for its intent, probably using an LLM. This LLM then generates a "good retrieval query" based on that intent.

Example: Most people's medical questions.

**Cost-Benefit Tradeoff**

Are these or other additional processing steps worth it for your application? They'll probably add complexity, processing costs, and user-experienced latency. Is the improvement in performance worth these costs?

### Vector Databases

What I skimmed over when performing a semantic search: the knowledge base is typically converted into a **vector database**.

Vector DB = a special type of database optimized for vector search.

-   They store and search over high-dimensional vector embeddings, which are numerical representations for unstructured data (e.g., text and images).
-   A similarity search becomes a computation of the distance between the query's vector embedding and a document's/chunk's vector embedding.
-   Almost synonymous with RAG systems.
