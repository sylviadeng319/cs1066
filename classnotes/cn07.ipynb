### Object-Oriented Programming

Python is an **object-oriented (OO)** programming language. You accepted this when you didn't question that everything in Python is an object. Briefly,

-   An OO programming language organizes code around objects instead of actions (i.e., commands or functions).
-   An object has attributes (data fields) and methods (functions that work within the context of an object).
-   A **class** is a template for creating a specific type of object. It allows you to create new data types.

`kings.py` creates a data type called `Hand`. See lines 4-30.

-   It has two data attributes: `name`, which identifies whose hand it is; and `hand`, which is a list of cards. See lines 7-8.
-   The rest of the class defines methods that the program can perform on an object of type `Hand`. For example, you can append a card to this hand (see lines 24-26).

In Python, method names that start and end with two underbar characters (`_`) are called **magic methods**, and they allow you to have Python treat your data types just like any built-in data type. 

-   For example, `__len__` is called behind the scenes if you passed an object of type `Hand` to the built-in function `len`. The implementation of the `__len__` method says that this operation should return the length of the `hand` list in this `Hand` object. `self` is the object whose length you want.

    ```python
    player1 = Hand('Human')  # creates a Hand object and names it player1
    len(player1)             # becomes __len__(player1) ==> len(player1.hand)
    ```

-   `__init__` is the **constructor**, which builds an instance of type `Hand`. It's what's executed for the statement `player1 = Hand('Human')`.

-   `__add__` is the wiring so that you can use the overloaded `+` operator with your data type.

> Don't worry if you don't follow all this jargon. We'll play with OO concepts in section this week.

> OO techniques work well when used judiciously. Notice that I didn't create a new data type for the playing cards in this game. It was overkill for something that could be a simple string. Of course, had I done so, I probably would have avoided the bug we just solved.

### Splitting the Input Space for Efficient Testing

In testing the function `is_even`, we naturally gravitated toward treating even and odd integers as two different test cases. It didn't matter what even number we tested, and it wasn't sufficient to just test even numbers.

Let's apply this idea of **equivalence classes** to a slightly more complicated function:

````python
def largest(a, b, c):
    if a > b and a > c:
        return a
    elif b > a and b > c:
        return b
    else:
        return c
````

**Question:** Using the formal parameter names (`a`, `b`, and `c`), what are the interesting categories to test?

-   You can ignore any values that are exceptional. In other words, `a`, `b`, and `c` are all the same type of object, and this object defines the `>` operator. E.g., strings and ints.

### Python's assert Statement

The pattern that we want `assert` to check:

````python
# assert actual_result == expected_result
assert largest(1, 2, 3) == 3     # A test case
assert largest(10, 10, 3) == 10  # Another test case
````

A failed test is useful information, and while it is nice to see lots of green checkmarks for passed tests, we really want to know about the failed tests. `assert` can help.

-   When the asserted condition is `True`, `assert` is silent.
-   When the asserted condition is `False`, `assert` throws an `AssertionError`.

You can add an optional message that's printed when an `AssertionError` occurs. I'll demonstrate the difference between `assert False` and `assert False, 'Failed Test #1'` in the interactive Python interpreter.

### Regression Tests

We've focused on building **unit tests**, which are pieces of code that verify the behavior of a function or other slice of a program.

**Regression testing** is the practice of running a bank of unit tests (and possibly other types of tests) after changes have been applied to a code base to verify that what you expect to still work does continue to work. You'll get a flavor for this in the pset.

Whether you need regression tests for your prototyping work is hard to say definitively. You will want some kind of regression suite if you find yourself relying on a tool and making changes to it over time.
