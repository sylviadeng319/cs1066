### Terminology

**Large Language Model (LLM)**

-  Neural network with many layers and some "secret sauce" weights, tuned by many training runs.
-  A model answers a prompt (of words) with a likely continuation of words.
-  An LLM is the "intelligence" engine, but it can only talk.

**A Harness**

-  The software wrapped around a model that allows it to do things we care about _that require intelligence_.
-  A harness contains things like a facility for memory, managing context, testing things in a sandbox, connecting to other interfaces (e.g., search the web for data or information), etc.
-  You've been using a harness + a model in the Codespace IDE. ChatGPT is a harness + a model.

**Agent** = AI model + harness code

-  There's value in the model _and_ the harness.
-  Who says that coders or tech moguls are the only people who should create the harness you need for your work?
-  You can build your own harness.

> In section, you'll take our web-browser-based Kings game with a dumb AI player, connect it to the API for an AI model, and create a smart AI player.

### Codespace Harness in Action

If you pay attention to what's in your codespace's AI response, you'll see the harness running what's called an **agent loop**, which is essentially a problem-solving loop.

1.  Read what's in the context and prompt
2.  Design then decompose into a task list
3.  Select a task to implement
4.  Implement selected task
5.  Check the result against the task goal
6.  Return to (4) if fail, (3) if pass, (2) if need new approach

The harness and its loop allow AI to:

-   "Think step-by-step" (problem decomposition).
-   Try multiple solutions (design thinking).
-   Check each solution, probably against a standard or goal (testing).
-   Keep the best work and redo sub-par work (apply judgment).
-   Move from answering a question to completing a complex task.
-   Look intelligent despite the "brain" being just a prediction engine.

Further reading: https://read.technically.dev/p/whats-harness-engineering

### Strategies Around Rate Limits

Google sets limits on your API calls and use of the Gemini AI model. You should pay attention to these limits.

> https://ai.google.dev/gemini-api/docs/pricing

In section, you'll build an AI player for the Kings game. Here are some tips to make the most of your limited free prompts:

-   As much as possible, test your program's functions and functionality without making AI API calls (e.g., use the 2-human player mode to test game play).

-   You'll discover that you'll want to tune the prompt you send in the API call (e.g., what context do you want in the prompt?). Tune your prompt using testing mode and a chatbot.

-   One of the rate-limiting metrics is "requests per minute." You shouldn't hit this limit as the Kings game alternates between human and AI. In other applications, you might need to slow down your rate of requests using the `time.sleep` function and protect against a poorly performing AI that eats up all your free credits.

-   There are several Gemini models that generate text. If you hit a rate limit with one model, switch to a different one. Or, in later psets, take a break and come back when your allowance has reset!
