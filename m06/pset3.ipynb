### Extend the Game

If this was your game, you would (and probably should) ask the AI to generate tests for the game's other important functions. You'd learn that pytest is capable of dealing with user input and randomness. You could use it to test the playing of several game turns.

But this pset goes in a different direction: Let's use AI to make a modification to the game and see if it breaks the tests you just built. This is regression testing.

Please follow these steps:

1.  Prompt the codespace AI with this text: "Modify `cards.py` to include a single joker card. Then modify `kings.py` so that the game treats this joker as a wildcard. Its rank can match any other rank. For counting points, the joker has a value of 10 points."

2.  Read the AI's response. It should make modifications to `cards.py`, which should be easy to check and accept. It will make modifications to `kings.py`, which will be much harder to check visually. The new logic for `drop_3` is even more complex than previously. This is when we'll rely on our tests.

3.  Accept the AI edits to `cards.py` and `kings.py`.

4.  Did the AI also mention that it extended the tests? If it didn't, ask it to. Then,

    -  Open the tests file.
    -  Read through the changes. The AI should not have touched the original tests. It should have just added new ones. This is exactly the behavior that we'd like to see: Our change didn't affect any of the old game rules and these tests are still valid.

5.  Accept the AI edits to the tests file.

**Q6:** Reflect on the work that the AI did to extend the game and test its extension. Are you satisfied with the result? Did anything suprise you about its implementation? Would you have characterized the test space differently?
